```python


import sys
import os

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns 


sys.path.append(os.path.abspath('..'))

from src import ROW_DATA_DIR , CATEGORICAL_COLUMNS , loader , NUMERICAL_COLUMNS
df =  loader()
```

```python
df.head()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.countplot(data=df, x='gender', ax=axes[0])
axes[0].set_title('Gender Distribution')
axes[0].set_xlabel('Gender')


sns.countplot(data=df, x='SeniorCitizen', ax=axes[1])
axes[1].set_title('Senior Citizen Distribution')
axes[1].set_xlabel('Senior Citizen')

sns.countplot(data=df, x='Partner', ax=axes[2])
axes[2].set_title('Partner Distribution')
axes[2].set_xlabel('Partner')

plt.tight_layout()

```

```python
fig , axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(data=df, x='tenure', bins=30, kde=True, ax=axes[0])
plt.axvline(df['tenure'].mean(), color='r', linestyle='--', label='Mean')
plt.axvline(df['tenure'].median(), color='g', linestyle='-', label='Median')
axes[0].set_title('Tenure Distribution')
axes[0].set_xlabel('Tenure (months)')

sns.histplot(data=df, x='MonthlyCharges', bins=30, kde=True, ax=axes[1])
plt.axvline(df['MonthlyCharges'].mean(), color='r', linestyle='--', label='Mean')
plt.axvline(df['MonthlyCharges'].median(), color='g', linestyle='-', label='Median')
axes[1].set_title('Monthly Charges Distribution')
axes[1].set_xlabel('Monthly Charges ($)')

sns.histplot(data=df, x='TotalCharges', bins=30, kde=True, ax=axes[2])
plt.axvline(df['TotalCharges'].mean(), color='r', linestyle='--', label='Mean')
plt.axvline(df['TotalCharges'].median(), color='g', linestyle='-', label='Median')
axes[2].set_title('Total Charges Distribution')
axes[2].set_xlabel('Total Charges ($)') 
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))





sns.histplot(data=df, x='tenure' , hue='Churn' , element='step' , fill=False , binwidth=2, ax=axes[0] ,kde=True)
axes[0].set_title('Tenure Distribution')
axes[0].set_xlabel('Tenure (months)')
axes[0].set_ylabel('Count')

sns.histplot(data=df, x='MonthlyCharges' , hue='Churn' , element='step' , fill=False , binwidth=4 , ax=axes[1] , kde=True)
axes[1].set_title('Monthly Charges Distribution')
axes[1].set_xlabel('Monthly Charges')
axes[1].set_ylabel('Count')

sns.histplot(data=df, x='TotalCharges' , hue='Churn' , element='step' , fill=False ,  common_norm=True  , binwidth=200 , ax=axes[2], kde=True,)
axes[2].set_title('Total Charges Distribution')
axes[2].set_xlabel('Total Charges')
axes[2].set_ylabel('Count')

plt.tight_layout()


```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

sns.boxplot(data=df, x='tenure', ax=axes[0])
axes[0].set_title('Tenure Boxplot')
axes[0].set_xlabel('Tenure (months)')

sns.boxplot(data=df, x='MonthlyCharges', ax=axes[1])
axes[1].set_title('Monthly Charges Boxplot')
axes[1].set_xlabel('Monthly Charges')

sns.boxplot(data=df, x='TotalCharges', ax=axes[2])
axes[2].set_title('Total Charges Boxplot')
axes[2].set_xlabel('Total Charges')


plt.tight_layout()

```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.scatterplot(
    data=df, 
    x='MonthlyCharges',
    y='tenure',
    hue='gender',
    s=25, 
    ax=axes[0]
)

plt.title('Monthly Charges by Tenure and Gender')
plt.xlabel('Monthly Charges')
plt.ylabel('Tenure')


sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=df,
    x='TotalCharges',
    y='tenure',
    hue='gender',
    s=25, 
    ax=axes[1]
)

plt.title('Total Charges by Tenure and Gender')
plt.xlabel('Total Charges')
plt.ylabel('Tenure')
plt.show()


plt.tight_layout()
```
